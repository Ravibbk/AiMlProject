```python
import gymnasium as gym
from gymnasium import spaces
import numpy as np
from stable_baselines3 import PPO
from stable_baselines3.common.env_checker import check_env

class CanteenQueueEnv(gym.Env):
    """
    A custom environment for Canteen Queue Management
    """
    def __init__(self):
        super(CanteenQueueEnv, self).__init__()
        
        # System Limits
        self.max_queue = 100       # Maximum people that can fit in the queue
        self.max_counters = 5      # Maximum number of food counters available
        self.service_rate = 3      # Each open counter serves 3 students per minute
        self.max_steps = 60        # 60 minutes (1 hour lunch rush)
        
        # Action Space: 3 actions
        # 0: Close 1 counter, 1: Do nothing, 2: Open 1 counter
        self.action_space = spaces.Discrete(3)
        
        # Observation Space: What the AI sees [current_queue_length, open_counters]
        self.observation_space = spaces.Box(
            low=np.array([0, 1]), 
            high=np.array([self.max_queue, self.max_counters]), 
            dtype=np.int32
        )

    def reset(self, seed=None, options=None):
        super().reset(seed=seed)
        
        self.current_queue = 0
        self.open_counters = 1     # Always start with at least 1 counter open
        self.current_step = 0
        
        return self._get_obs(), {}

    def _get_obs(self):
        return np.array([self.current_queue, self.open_counters], dtype=np.int32)

    def _get_arrival_rate(self):
        # Simulates a "lunch rush" curve. 
        # Peaks around step 30 (middle of the hour)
        if 20 <= self.current_step <= 40:
            return np.random.poisson(lam=15) # High traffic (avg 15 students/min)
        else:
            return np.random.poisson(lam=3)  # Low traffic (avg 3 students/min)

    def step(self, action):
        self.current_step += 1
        
        # 1. Apply Agent's Action (Open/Close Counters)
        if action == 0 and self.open_counters > 1:
            self.open_counters -= 1  # Close a counter
        elif action == 2 and self.open_counters < self.max_counters:
            self.open_counters += 1  # Open a counter
            
        # 2. Simulate Environment Dynamics
        # Students arrive
        arrivals = self._get_arrival_rate()
        self.current_queue = min(self.max_queue, self.current_queue + arrivals)
        
        # Students are served
        served = self.open_counters * self.service_rate
        self.current_queue = max(0, self.current_queue - served)
        
        # 3. Calculate Reward (The core logic)
        # We heavily penalize long queues, but also penalize having too many counters open
        wait_penalty = self.current_queue * 2 
        staff_cost = self.open_counters * 5   
        
        reward = -(wait_penalty + staff_cost)
        
        # 4. Check if episode is over
        truncated = False
        if self.current_step >= self.max_steps:
            truncated = True
            
        return self._get_obs(), reward, False, truncated, {}


# --- Training and Execution ---
if __name__ == "__main__":
    # 1. Create and verify the environment
    env = CanteenQueueEnv()
    check_env(env, warn=True)
    print("Environment verified! Starting training...\n")

    # 2. Initialize the AI Model
    model = PPO("MlpPolicy", env, verbose=0)

    # 3. Train the AI (100,000 minutes of simulation)
    print("Training AI for 100,000 steps (This will take a few seconds)...")
    model.learn(total_timesteps=100000)
    print("Training complete!\n")

    # 4. Test the trained AI on a simulated Lunch Rush
    print("--- Simulating 1-Hour Lunch Rush ---")
    obs, info = env.reset()
    
    total_reward = 0
    for minute in range(60):
        action, _states = model.predict(obs, deterministic=True)
        obs, reward, terminated, truncated, info = env.step(action)
        total_reward += reward
        
        queue_size = obs[0]
        counters = obs[1]
        
        # Print status every 10 minutes to see how the AI adapts
        if minute % 10 == 0 or minute == 59:
            rush_status = "LUNCH RUSH!" if 20 <= minute <= 40 else "Normal traffic"
            print(f"Minute {minute:02d} | Queue: {queue_size:02d} | Open Counters: {counters} | Event: {rush_status}")
            
    print(f"\nSimulation Finished. Total Score: {total_reward}")
```
